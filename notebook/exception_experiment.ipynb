```python
import sys
import traceback
try:
    # Code that might cause an error
    x = 10 / 0

except Exception as e:
    # Runs if an error occurs
    print(e)
    _,_,trc_back=sys.exc_info()
    filename=trc_back.tb_frame.f_code.co_filename
    print(filename)
    lineno=trc_back.tb_lineno
    print(lineno)
    error_msg=str(e)
    print(error_msg)
    
    traceback_syc=''.join(traceback.format_exception(*sys.exc_info()))
    print(traceback_syc)
```
